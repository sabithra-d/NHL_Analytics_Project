```python
import pymysql
from sqlalchemy import create_engine
engine = create_engine('mysql+pymysql://root:sabi3601@localhost:3306/nhl_analytics') #create engine to connect mysql db
```

```python
#run this in db
USE nhl_analytics;

create table game_stats(
stat_id INT AUTO_INCREMENT PRIMARY KEY,
game_id BIGINT,
player_id BIGINT,
team_id INT,
goals INT,
assists INT,
points INT,
shorts_on_goal INT,
penalty_min INT,
toi VARCHAR(10),
plus_minus INT,
FOREIGN KEY (player_id) REFERENCES players(player_id),
FOREIGN KEY (team_id) REFERENCES teams(team_id))

#use this to craete table with all constrints and run below for data insertion. 
```

```python
df_games.to_sql('game_stats', con=engine, if_exists='replace') #push the data inside sql table game_stats. if_exists='replace' means if table already exist then replace it with new data. 
#connect to sql, create table as "game_stats" with the engine we craeted above, if tabel already exist do replace.
#didnt craete any table or insert data manually.
#pushing data from json file to game_stats table in sql.easiest way.or else use for loop method
```

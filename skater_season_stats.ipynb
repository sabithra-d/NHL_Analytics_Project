```python
import pymysql
from sqlalchemy import create_engine
engine = create_engine('mysql+pymysql://root:sabi3601@localhost:3306/nhl_analytics')
```

```python
#run this in db
USE nhl_analytics;

create table skater_season_stats(
stat_id INT AUTO_INCREMENT PRIMARY KEY,
player_id BIGINT,
season VARCHAR(20),
team_id INT,
games_played INT,
goals INT,
assists INT,
points INT,
plus_minus INT,
penalty_min INT,
shots INT,
avg_toi VARCHAR(10),
FOREIGN KEY (player_id) REFERENCES players(player_id),
FOREIGN KEY (team_id) REFERENCES teams(team_id))
```

```python
df_skater.to_sql('skater_season_stats', con=engine, if_exists='replace') #push the data inside sql table skater_season_stats. if_exists='replace' means if table already exist then replace it with new data. 
```

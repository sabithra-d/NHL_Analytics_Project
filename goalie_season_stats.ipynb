```python
import pymysql
from sqlalchemy import create_engine
engine = create_engine('mysql+pymysql://root:sabi3601@localhost:3306/nhl_analytics')
```

```python
#run this in db
USE nhl_analytics;

create table goalie_season_stats(
stat_id INT AUTO_INCREMENT PRIMARY KEY,
player_id BIGINT,
season VARCHAR(20),
team_id INT,
games_played INT,
wins INT,
losses INT,
ot_losses INT,
save_pct FLOAT,
goals_against_avg FLOAT,
shutouts INT,
saves INT,
FOREIGN KEY (player_id) REFERENCES players(player_id),
FOREIGN KEY (team_id) REFERENCES teams(team_id))
```

```python
df_goalie.to_sql('goalie_season_stats', con=engine, if_exists='replace') #push the data inside sql table goalie_season_stats. if_exists='replace' means if table already exist then replace it with new data. 
```

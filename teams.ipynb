```python
import requests

url = "https://api-web.nhle.com/v1/standings/now"

response = requests.get(url)

response.status_code # 200 indicates success
```

```python
data = response.json()
```

```python
data.keys()
```

```python
len(data['standings'])
```

```python
teams = []

for team in data['standings']:
    teams.append({
        'team_abbrev': team['teamAbbrev']['default'],
        'team_name': team['teamName']['default'],
        'conference_name': team['conferenceName'],
        'division_name': team['divisionName'],
        'logo_url': team['teamLogo']
    })

teams
```

```python
#connection string
import pymysql

#create a connection
conn = pymysql.connect(
    host = 'localhost',
    user = 'root',
    password = "sabi3601",
    database = 'nhl_analytics'
)

#create a cursor
cursor = conn.cursor() #this is how we make sure the connections
print("Connected to NHL Analytics database")
```

```python
create_teams_table = """
CREATE TABLE IF NOT EXISTS teams (
    team_id INT AUTO_INCREMENT PRIMARY KEY,
    team_abbrev VARCHAR(10) NOT NULL UNIQUE,
    team_name VARCHAR(100) NOT NULL,
    conference_name VARCHAR(50),
    division_name VARCHAR(50),
    logo_url TEXT
)
"""

cursor.execute(create_teams_table)

conn.commit()

print("Teams table created successfully")
```

```python
insert_query = """INSERT INTO teams (team_abbrev,team_name,conference_name,division_name,logo_url)
                    VALUES(%s,%s,%s,%s,%s)"""

for i in teams:
   values = (i['team_abbrev'],
         i['team_name'],
         i['conference_name'],
         i['division_name'],
         i['logo_url'])
   cursor.execute(insert_query, values)   # <-- indented inside the loop

conn.commit()

print(f"{len(teams)} teams inserted successfully")
```
